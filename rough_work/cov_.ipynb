```python
import numpy as np
```

```python
study_hours = np.array([2, 4, 6, 8, 10])
exam_scores = np.array([55, 60, 75, 85, 95])
```

```python
mean_study_hours = np.mean(study_hours)
mean_exam_scores = np.mean(exam_scores)
print(f"Mean of study hours = {mean_study_hours}")
print(f"Mean exam scores = {mean_exam_scores}")
```

```python
def calc_variance(A):
    mean_A = np.mean(A)
    devs = (A - mean_A) ** 2
    return np.sum(devs) / (A.shape[0] - 1)
```

```python
def calc_devs(A, B):
    mean_A = np.mean(A)
    mean_B = np.mean(B)
    dev_A = A - mean_A
    dev_B = B - mean_B
    return np.dot(dev_A, dev_B) / (len(A) - 1)
```

```python
devs_exam_scores = calc_variance(exam_scores)
print(devs_exam_scores)
devs_study_hours = calc_variance(study_hours)
print(devs_study_hours)
dev_study_exam = calc_devs(study_hours, exam_scores)
print(dev_study_exam)
```
